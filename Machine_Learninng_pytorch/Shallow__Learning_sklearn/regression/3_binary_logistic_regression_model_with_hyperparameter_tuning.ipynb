# BINARY LOGISTIC REGRESSION WITH HYPERPARAMETER TUNING

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
from sklearn.datasets import make_classification
```

```python
# create dataset
X, y = make_classification(n_samples=1000, n_features=10, n_classes=2, random_state=42)
```

```python
df = pd.DataFrame(X)
df, y
```

```python
# trai n test split
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(X, y, test_size=0.30, random_state=42)
y_test
```

# MODEL TRAINING

```python
from sklearn.linear_model import LogisticRegression
bin_logi_reg_model = LogisticRegression()
```

```python
bin_logi_reg_model.fit(x_train,y_train)
```

```python
y_pred = bin_logi_reg_model.predict(x_test)
```

```python
y_pred
```

```python
y_prob = bin_logi_reg_model.predict_proba(x_test)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
accuracy_score = accuracy_score(y_test,y_pred)
accuracy_score
```

```python
confusion_matrix = confusion_matrix(y_test,y_pred)
confusion_matrix
```

```python
classification_report = classification_report(y_test,y_pred, output_dict=True)
classification_report
```

# HYPERPARAMETER TUNING

### 1)GRID SEARCH HYPERPARAMETER TUNING

```python
model = LogisticRegression()
penalty = ['l1', 'l2', 'elasticnet']
c = [100,10,1.0,0.1,0.01]
solver = ['newton-cg', 'lbfgs', 'liblinear', 'sag', 'saga']
```

```python
params = dict(penalty=penalty, C=c, solver=solver,)
params
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.model_selection import StratifiedKFold
```

```python
cv = StratifiedKFold(n_splits=5, shuffle=False)
grid = GridSearchCV(estimator=model, param_grid=params, scoring='accuracy', cv=cv, n_jobs=-1)
```

```python
grid
```

```python
grid.fit(x_train, y_train)
```

```python
grid.best_params_
```

```python
grid.best_score_
```

```python
y_pred = grid.predict(x_test)
y_pred
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
accuracy_score = accuracy_score(y_test,y_pred)
accuracy_score
```

```python
confusion_matrix = confusion_matrix(y_test,y_pred)
confusion_matrix
```

```python
classification_report = classification_report(y_test,y_pred, output_dict=True)
classification_report
```

### 2)RANDOMISED SEARCH CV

```python
from sklearn.model_selection import RandomizedSearchCV
```

```python
model = LogisticRegression()
random_cv = RandomizedSearchCV(estimator=model, param_distributions=params, cv=5, scoring='accuracy')
```

```python
random_cv.fit(x_train,y_train)
```

```python
random_cv.best_score_
```

```python
random_cv.best_params_
```

```python
y_pred = random_cv.predict(x_test)
y_pred
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
accuracy_score = accuracy_score(y_test,y_pred)
accuracy_score
```
