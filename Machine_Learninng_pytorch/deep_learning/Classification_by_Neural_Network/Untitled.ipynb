### 1) transform data/images into tensors

```python
# image resize + data augmentation + images to tensors
data_transformer = transforms.Compose([
    transforms.Resize(size=(64,64)),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.ToTensor()]
)
```

```python
from PIL import Image
image = next(images_list)
parentfolder = image.parent.name
pil_image = Image.open(image)
data_transformer(pil_image)
parentfolder
```

```python
transformerd_image = data_transformer(pil_image)
transformerd_image.shape
```

```python
# visualize
plt.imshow(pil_image)
plt.title(label = f'Class:{parentfolder}')
```
