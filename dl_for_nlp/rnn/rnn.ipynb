# Basic RNN 

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing import sequence
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, Dense, SimpleRNN
```

```python
# load the inbuilt dataset
vocab_size = 1000
# (x_train, y_train),(x_test, y_test) = imdb.load_data(path="/home/aj/GenAI/dl_for_nlp/rnn/Data",  num_words=vocab_size)
(x_train, y_train),(x_test, y_test) = imdb.load_data(num_words=vocab_size)

# print(f"x_train={x_train.shape} \n y_train={y_train.shape} \n x_test={x_test.shape} \n y_test={y_test.shape}")
```

```python
# Sequencing (for uniform input to neuron)

# check reviews lengths
# for review in x_train:
#     print(len(review))

# we need to make our all reviews of same length
x_train = sequence.pad_sequences(x_train, maxlen=500, padding='pre')
x_test = sequence.pad_sequences(x_test, maxlen=500, padding='pre')
# x_train
# len(x_train)
```

# RNN MOdelling

```python
# sequential model
rnn_model = Sequential()

rnn_model.add(Embedding(input_dim=1000, output_dim=300, input_length=500))
rnn_model.add(SimpleRNN(units=128, activation='relu'))
rnn_model.add(Dense(units=1, activation='sigmoid'))
```

```python
rnn_model.summary()
```

```python
rnn_model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
## Create an instance of EarlyStoppping Callback
from tensorflow.keras.callbacks import EarlyStopping
earlystopping=EarlyStopping(monitor='val_loss',patience=5,restore_best_weights=True)
earlystopping
```

```python
x_train
```

```python
history=rnn_model.fit(
    x_train,y_train,epochs=10,batch_size=32,
    validation_split=0.2,
    callbacks=[earlystopping]
)
```

```python
## Save model file
rnn_model.save('simple_rnn_imdb.h5')
```
